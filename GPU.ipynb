```python
import torch
import time

# Matrix size
N = 2000

# CPU test
a = torch.randn(N, N)
b = torch.randn(N, N)
start = time.time()
c = torch.mm(a, b)
end = time.time()
print("CPU time:", end - start)

# GPU test
a = a.cuda()
b = b.cuda()
torch.cuda.synchronize()
start = time.time()
c = torch.mm(a, b)
torch.cuda.synchronize()
end = time.time()
print("GPU time:", end - start)
```
